Load environment variables

```python
from dotenv import load_dotenv

load_dotenv()  # Load environment variables from .env file
```

Create an API Client

```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-5-5"  # Replace with the desired model name
```

Make a request

```python
message = client.messages.create(
  model = model,
  max_tokens = 1000,
  messages = [
    {
      "role": "user",
      "content": "What is a quantum computing? Answer in one sentence."
    }
  ]
)
```

Print out the `message` variable

```python
message.content[0].text
```

Another message with no context

```python
message_out_of_context = client.messages.create(
  model = model,
  max_tokens = 1000,
  messages = [
    {
      "role": "user",
      "content": "Write another sentence."
    }
  ]
)
```

Print out the second message responce

```python

text = "".join(b.text for b in message_out_of_context.content if b.type == "text")
print(text)
```
