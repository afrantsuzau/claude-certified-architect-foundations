```python
from dotenv import load_dotenv
from anthropic import Anthropic
from IPython.display import Markdown, display

load_dotenv()  # Load environment variables from .env file

client = Anthropic()  # Create a client to interact with the API
model = "claude-haiku-5-5"
messages = []  # Initialize an empty list to hold the conversation messages

def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})
    
def add_assistant_message(messages, content):
    messages.append({"role": "assistant", "content": content})
    
def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }
    
    if system is not None:
        params["system"] = system
    
    message = client.messages.create(**params)
    
    return "".join(b.text for b in message.content if b.type == "text")

def print_markdown(text):
    display(Markdown(text))
```

```python
system = """
            You are a Python developer who provides clear and very concise code solutions.
        """

add_user_message(messages, "Write a Python function that checks a string for duplicate characters and returns True if duplicates are found, otherwise returns False.")
answer = chat(messages, system=system)
print_markdown(answer)
```
